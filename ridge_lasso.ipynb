# algerian fire forest dataset

this dataset includes 244 instances that regroup a data of two regions of algeria namely bajaia region located in northeast of algeria and sidi-bel-abbes region located in northwest of algeria.
122 instances for each region.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

```

```python
df=pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv' , header=1)
```

```python
df.head()
```

```python
df.info()
```

```python
df[df.isnull().any(axis=1)]


```

the dataset is converted into two sets based on region fromm 122th index , we can make new column based on region 
1. Bejaia region dataset
2. sidi-bel-abbes region



```python
df.loc[:122,'region']=0
df.loc[122:,'region']=1

df.info()





```

```python
### converting float values to int

df[['region']]= df[['region']].astype(int)
```

```python
df.head()
```

```python
df.tail()
```

```python
df.isnull().sum()
```

```python
## removing 122 rows from dataset
df=df.dropna().reset_index(drop=True)
df.head()




```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
## droping 122 row

df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
## removing extra space in column names
df.columns=df.columns.str.strip()
df.columns



```

```python
df.info()





```

```python
## changes the required columns as int datatypes
df[['day', 'month', 'year', 'Temperature', 'RH']] = df[['day', 'month', 'year', 'Temperature', 'RH']].astype(int)
```

```python
df.info()
```

```python
## changing to float data types
objects=[features for features in df.columns if df[features].dtype == 'str']
```

```python
for i in objects:
    if i!= 'Classes':
        df[i]=df[i].astype(float)
        
```

```python
df.info()
```

```python
## lets save cleaned dataset
df.to_csv('Algerian_forest_fires_cleaned_dataset.csv' , index=False)
```

```python
## ploting pie chart

from cProfile import label


classlabel = ["Fire","Not Fire "]
plt.figure(figsize=(12,7))

# Assign blue to the maximum value (most samples), red to the minimum value

colors = ['blue' if p == percentage.max() else 'red' for p in percentage]

plt.title("pie chart of dataset")

plt.pie(percentage, labels=classlabel, autopct='%1.1f%%', colors=colors)
```
