```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
x= 6* np.random.rand(100,1) - 3
y= 0.5*x**2 + x + 2 + np.random.randn(100,1)
##quadratic equation[ y= 0.5x^2 + x + 2 + outliers ]

plt.scatter(x,y,color='red')
plt.xlabel('x')
plt.ylabel('y')
```

```python
from sklearn.model_selection import train_test_split
```

```python


x_train,x_test,y_train,y_test=train_test_split(x,y, test_size=0.2 ,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression

regressor=LinearRegression()
regressor.fit(x_train,y_train)


```

```python
from sklearn.metrics import mean_squared_error, r2_score

score = r2_score(y_test,regressor.predict(x_test))
print(score)
```

```python
plt.plot(x_train,regressor.predict(x_train),color='blue')
plt.scatter(x_train,y_train,color='red')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.title('Training set')
plt.show()
```

```python
from sklearn.preprocessing import PolynomialFeatures

poly=PolynomialFeatures(degree=2,include_bias=True)

x_train_poly=poly.fit_transform(x_train)
x_test_poly=poly.fit_transform(x_test)
```

```python
x_train_poly
```

```python
from sklearn.metrics import mean_squared_error, r2_score
regressor=LinearRegression()
regressor.fit(x_train_poly,y_train)


y_pred=regressor.predict(x_test_poly)

score=r2_score(y_test,y_pred)

print(score)
```

```python
print(regressor.coef_)
```

```python
print(regressor.intercept_)
```

```python
plt.scatter(x_train,regressor.predict(x_train_poly),color='red')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.title('Training set')
plt.show()
plt.scatter(x_train,y_train)
```

```python
from sklearn.preprocessing import PolynomialFeatures


poly=PolynomialFeatures(degree=3,include_bias=True)

x_train_poly=poly.fit_transform(x_train)
x_test_poly=poly.fit_transform(x_test)
```

```python
x_train_poly
```

```python
from sklearn.metrics import mean_squared_error, r2_score
regressor=LinearRegression()
regressor.fit(x_train_poly,y_train)


y_pred=regressor.predict(x_test_poly)

score=r2_score(y_test,y_pred)

print(score)
```

```python
## prediction for a new data
x_new=np.linspace(-3,3,200).reshape(200,1)

x_new_poly=poly.fit_transform(x_new)
```

```python
x_new_poly
```

```python
y_new=regressor.predict(x_new_poly)
plt.plot(x_new,y_new,'r-',label='newPredictions')
plt.plot(x_train,y_train,'b.',label='Training data')
plt.plot(x_test,y_test,'g.',label='Testing data')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.legend()
plt.show()
```

```python
from sklearn.pipeline import Pipeline

x_new=np.linspace(-3,3,200).reshape(200,1)

poly_features=PolynomialFeatures(degree=2,include_bias=True)

line_regression=LinearRegression()

poly_regression=Pipeline([
    ('poly_features',poly_features),
    ('line_regression',line_regression)
    ])
## polynomial aand fit of linear regression
poly_regression.fit(x_train,y_train)

y_new_poly=poly_regression.predict(x_new)

## plotting the polynomial regression

plt.plot(x_new,y_new_poly,'r-',label='Polynomial Predictions')
plt.plot(x_train,y_train,'b.',label='Training data')
plt.plot(x_test,y_test,'g.',label='Testing data')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.axis([-4,4,0,10])
plt.show()
```
